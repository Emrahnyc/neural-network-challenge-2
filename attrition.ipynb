## Part 1: Preprocessing

```python
# Import our dependencies
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
import pandas as pd
import numpy as np
from tensorflow.keras.models import Model
from tensorflow.keras import layers

#  Import and read the attrition data
attrition_df = pd.read_csv('https://static.bc-edx.com/ai/ail-v-1-0/m19/lms/datasets/attrition.csv')
attrition_df.head()
```

```python
# Determine the number of unique values in each column
attrition_df.nunique()
```

```python
# Create y_df with the Attrition and Department columns
y_df= attrition_df[['Attrition', 'Department']]
# 
```

```python
y_df["Department"].value_counts()
```

```python
y_df["Attrition"].value_counts()
```

```python
attrition_df.columns
```

```python
# Create a list of at least 10 column names to use as X data
list_of_columns = ['Age', 'DistanceFromHome',
       'Education', 'JobSatisfaction', 'NumCompaniesWorked', 'OverTime', 'StockOptionLevel','WorkLifeBalance',
       'YearsAtCompany', 'YearsSinceLastPromotion']
# Create X_df using your selected columns
X= attrition_df[list_of_columns]

# Show the data types for X_df
X.dtypes
```

```python
X.head()
```

```python
# Split the data into training and testing sets
from sklearn.model_selection import train_test_split
X_train, X_test,y_df_train,y_df_test = train_test_split(X, y_df)
```

```python
X_train.dtypes
```

```python
X_test.dtypes
```

```python
# Convert your X data to numeric data types however you see fit
# Add new code cells as necessary
overtime_encoder= OneHotEncoder(sparse_output=False, handle_unknown='ignore', drop='first')
overtime_encoder.fit(X_train[['OverTime']])
X_train["Overtime_Encoded"]= overtime_encoder.transform(np.array(X_train['OverTime']).reshape(-1, 1))
X_test["Overtime_Encoded"]= overtime_encoder.transform(np.array(X_test['OverTime']).reshape(-1, 1))
X_train= X_train.drop(columns=['OverTime'])
X_test= X_test.drop(columns=['OverTime'])

```

```python
X_train.head()
```

```python
# Create a StandardScaler

s_scaler= StandardScaler()
# Fit the StandardScaler to the training data
s_scaler.fit(X_train)


# Scale the training and testing data
X_train_scaled = s_scaler.transform(X_train)
X_test_scaled = s_scaler.transform(X_test)
```

```python
from sklearn.preprocessing import OneHotEncoder

# Create a OneHotEncoder for the Department column
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore', drop=None)

# Fit the encoder to the "Department" column in the training data
ohe.fit(np.array(y_df["Department"]).reshape(-1, 1))

# Create two new variables by applying the encoder to the training and testing data
y_scaled_train_department= ohe.transform(np.array(y_df_train['Department']).reshape(-1, 1))
y_scaled_test_department= ohe.transform(np.array(y_df_test['Department']).reshape(-1, 1))

```

```python
# Create a OneHotEncoder for the Attrition column

ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore', drop=None)

# Fit the encoder to the "Department" column in the training data
ohe.fit(np.array(y_df["Attrition"]).reshape(-1, 1))

# Create two new variables by applying the encoder to the training and testing data
y_scaled_train_attrition= ohe.transform(np.array(y_df_train['Attrition']).reshape(-1, 1))
y_scaled_test_attrition= ohe.transform(np.array(y_df_test['Attrition']).reshape(-1, 1))
```

```python
y_scaled_train_department
```

## Part 2: Create, Compile, and Train the Model

```python
# Find the number of columns in the X training data.
X_train_scaled.shape[1]

# Create the input layer
input = layers.Input(shape=(X_train_scaled.shape[1],), name='input_layer')
# Create at least two shared layers
shared1 = layers.Dense(64, activation='relu', name='second_layer')(input)
shared2 = layers.Dense(128, activation='relu', name='third_layer')(shared1)
```

```python
# Create a branch for Department
# with a hidden layer and an output layer

# Create the hidden layer
department_hidden = layers.Dense(32, activation='softmax', name='department_hidden')(shared2)

# Create the output layer
department_output = layers.Dense(3, activation='sigmoid', name='department_output')(department_hidden)
```

```python
# Create a branch for Attrition
# with a hidden layer and an output layer

# Create the hidden layer

attrition_hidden = layers.Dense(32, activation='softmax', name='attrition_hidden')(shared2)
# Create the output layer
attition_output = layers.Dense(2, activation='sigmoid', name='attrition_output')(attrition_hidden)
```

```python
# Create the model

model = Model(inputs=input, outputs=[department_output, attition_output])
# Compile the model
model.compile(optimizer='adam',
               loss={'department_output': 'categorical_crossentropy', 'attrition_output': 'binary_crossentropy'},
              metrics={'department_output': 'accuracy', 'attrition_output': 'accuracy'})

# Summarize the model
model.summary()
```

```python
# Train the model


model.fit(
    X_test_scaled,
    {'department_output': y_scaled_train_department, 'attrition_output': y_scaled_train_attrition},
    epochs=100,
    batch_size=32,
    validation_split=0.2
)
```

```python
# Evaluate the model with the testing data
test_results = model.evaluate(X_test_scaled, {'department_output': y_scaled_test_department,'attrition_output': y_scaled_test_attrition,})
test_results
```

```python
# Print the accuracy for both department and attrition
print(f"Department Accuracy: {test_results[3]}")
print(f"Attrition Accuracy: {test_results[4]}")
```
