## Part 1: Preprocessing

```python
# Import our dependencies
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np
from tensorflow.keras.models import Model
from tensorflow.keras import layers

#  Import and read the attrition data
attrition_df = pd.read_csv('https://static.bc-edx.com/ai/ail-v-1-0/m19/lms/datasets/attrition.csv')
attrition_df.head()
```

```python
# Determine the number of unique values in each column
attrition_df.nunique()
```

```python
# Create y_df with the Attrition and Department columns
y_df= attrition_df[['Attrition', 'Department']]
```

```python
y_df["Department"].value_counts()
```

```python
y_df["Attrition"].value_counts()
```

```python
attrition_df.columns
```

```python
# Create a list of at least 10 column names to use as X data
list_of_columns = ['Age', 'DistanceFromHome',
       'Education', 'JobSatisfaction', 'NumCompaniesWorked', 'OverTime', 'StockOptionLevel','WorkLifeBalance',
       'YearsAtCompany', 'YearsSinceLastPromotion']
# Create X_df using your selected columns
X_df = attrition_df[list_of_columns]

# Show the data types for X_df
X_df.dtypes
```

```python
X_df.head()
```

```python
# Split the data into training and testing sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_df, y_df)
```

```python
X_train.dtypes
```

```python
X_test.dtypes
```

```python
# Convert your X data to numeric data types however you see fit
# Add new code cells as necessary
X_df = X_df.apply(pd.to_numeric, errors='coerce')
```

```python
y_train.dtypes
```

```python
# Create a StandardScaler

scaler= StandardScaler()
# Fit the StandardScaler to the training data
scaler.fit(X_train)


# Scale the training and testing data
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.preprocessing import OneHotEncoder

# Create a OneHotEncoder for the Department column
ohe = OneHotEncoder(sparse_output=False)

# Fit the encoder to the "Department" column in the training data
y_scaler_department = ohe.fit(y_train[["Department"]])

# Create two new variables by applying the encoder to the training and testing data
y_train_encoded_department = y_scaler_department.transform(y_train[["Department"]])
y_test_encoded_department = y_scaler_department.transform(y_test[["Department"]])

```

```python
# Create a OneHotEncoder for the Attrition column

ohe= OneHotEncoder(sparse_output=False)
# Fit the encoder to the training data
y_scaler_attrition= ohe.fit(y_train[["Attrition"]])

# Create two new variables by applying the encoder
# to the training and testing data
y_train_encoded_attrition = y_scaler_attrition.transform(y_train[["Attrition"]])
y_test_encoded_attrition = y_scaler_attrition.transform(y_test[["Attrition"]])
```

## Part 2: Create, Compile, and Train the Model

```python
# Find the number of columns in the X training data.
X_train_scaled.shape[1]

# Create the input layer
input = layers.Input(shape=(X_train_scaled.shape[1],), name='input_layer')
# Create at least two shared layers
shared1 = layers.Dense(64, activation='relu', name='second_layer')(input)
shared2 = layers.Dense(128, activation='relu', name='third_layer')(shared1)
```

```python
# Create a branch for Department
# with a hidden layer and an output layer

# Create the hidden layer
department_hidden = layers.Dense(32, activation='softmax', name='department_hidden')(shared2)

# Create the output layer
department_output = layers.Dense(3, activation='sigmoid', name='department_output')(department_hidden)
```

```python
# Create a branch for Attrition
# with a hidden layer and an output layer

# Create the hidden layer

attrition_hidden = layers.Dense(32, activation='softmax', name='attrition_hidden')(shared2)
# Create the output layer
attition_output = layers.Dense(2, activation='sigmoid', name='attrition_output')(attrition_hidden)
```

```python
# Create the model

model = Model(inputs=input, outputs=[department_output, attition_output])
# Compile the model
model.compile(optimizer='adam',
               loss={'department_output': 'categorical_crossentropy', 'attrition_output': 'binary_crossentropy'},
              metrics={'department_output': 'accuracy', 'attrition_output': 'accuracy'})

# Summarize the model
model.summary()
```

```python
# Train the model


model.fit(
    X_test_scaled,
    {'department_output': y_test_encoded_attrition, 'attrition_output': y_test_encoded_department},
    epochs=100,
    batch_size=32,
    validation_split=0.2
)
```
